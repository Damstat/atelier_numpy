# Contexte 
Une entreprise dispose de plusieurs capteurs installés dans différents bâtiments. Les capteurs mesurent régulièrement la température, l'humidité, la pression et la consommation énergétique. 
Les données doivent être préparées et analysées avant d'être transmises à un futur système de Machine Learning permettant de détecter des situations anormales 


```python
# Importation de bibliothèque numpy
import numpy as np
```

# Partie 1 : Listes Python vs tableaux NumPy 

On dispose des températures mesurées par un capteur : 24.5, 25.1, 26.3, 27.0 et 25.8 

## 1)	Créer une liste Python stockant ces températures; 

```python
temperature = [24.5, 25.1, 26.3, 27.0, 25.8]
print("Temperature readings:", temperature)
```

## 2)	Créer un tableau NumPy équivalent stockant ces températures ; 

```python
temperature_tab = np.array(temperature)
temperature_tab
```

## 3)	Effectuer une multiplication par 2 sur ces deux collections ; 

```python
# multiplication par 2 pour la liste
temperature_X2 = temperature * 2
temperature_X2
```

# Partie 2 – Création des données 

## 1)	Créer un tableau représentant les mesures de température de 10 capteurs, initialement à zéro. 

```python
# Températures de 10 capteurs, initialement à zéro
temp_10_capteurs = np.zeros(10)
print(temp_10_capteurs)
```

## 2)Créer un tableau contenant 10 valeurs égales à 1. 

```python
# Tableau de 10 valeurs egales à 1
temp_10_capteurs_1 = np.ones(10)
print(temp_10_capteurs_1)
```

## 3)Créer un tableau de 10 valeurs initialisées à 25. 

```python
# Tableau de 10 valeurs initialisées à 25
temp_10_capteurs_25 = np.full(10, 25)
print(temp_10_capteurs_25)
```

## 4)Créer une matrice identité de taille 5 × 5. 

```python
# creation de la matrice identité 5x5
matrice_identite = np.eye(5)
print(matrice_identite)
```

## 5)Créer les 24 heures (de 0 à 23) de mesure d’une journée 

```python
# creation des 24 heures (de 0 à 23) de mesure d'une journée
heures = np.arange(24)
print(heures)
```

## 6)Créer 10 valeurs régulièrement réparties entre 20 °C et 30 °C  

```python
# 10 valeurs régulièrement réparties entre 20 et 30
temp_10_capteurs_20_30 = np.linspace(20, 30, 10)
print(temp_10_capteurs_20_30)
```

## 7)Créer aléatoirement les températures de 100 mesures, en rendant les résultats reproductibles 

```python
# creation aleatoirement des temperatures de 100 mesures, rendant les resultats reproductibles
np.random.seed(42)
temp_100_capteurs = np.random.uniform(20, 30, 100)
print(temp_100_capteurs)
```
