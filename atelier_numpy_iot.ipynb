# Partie 8 : Calcul scientifique 

Considérons le tableau représentant les 24 températures du 1er bâtiment. 

## 1)	Pour les températures calculer le minimum, le maximum, la somme, la moyenne, la médiane, la variance et l’écart-type 

```python
# 1) Statistiques descriptives
t_min = temp_jour.min()
t_max = temp_jour.max()
t_sum = temp_jour.sum()
t_mean = temp_jour.mean()
t_median = np.median(temp_jour)
t_var = temp_jour.var()
t_std = temp_jour.std()

print(f"Minimum       : {t_min:.2f}")
print(f"Maximum       : {t_max:.2f}")
print(f"Somme         : {t_sum:.2f}")
print(f"Moyenne       : {t_mean:.2f}")
print(f"Médiane       : {t_median:.2f}")
print(f"Variance      : {t_var:.2f}")
print(f"Écart-type    : {t_std:.2f}")
```

## 2)	Standardiser le tableau des températures puis vérifier les valeurs de la moyenne et de l’écarttype du tableau standardisé 

### Formule de la standardisation

$$
Z_i = \frac{X_i - \mu}{\sigma}
$$

```python
# standartisons le tableau des temperatures puis verifions les valeurs le la moyenne
temp_jour_standardiser = (temp_jour -  t_mean)/t_std

#Verifions les valeurs de la moyenne et de l'ecart-type
print("Tableau standardisé :", temp_jour_standardiser.round(2))
print("Moyenne après standardisation :", round(temp_jour_standardiser.mean(), 10))  # ~0
print("Écart-type après standardisation :", round(temp_jour_standardiser.std(), 10)) 
```

## 3)	Normaliser (Min-Max) le tableau des températures puis vérifier les valeurs du tableau normalisé

### Formule de la normalisation Min-Max

$$
X_{\text{normalisé}} =
\frac{X - X_{\min}}{X_{\max} - X_{\min}}
$$

Les valeurs normalisées sont comprises entre $0$ et $1$ :
$$
0 \leq X_{\text{normalisé}} \leq 1
$$

```python
# 3) Normalisation Min-Max : (x - min) / (max - min)
temp_jour_normalisees = (temp_jour - t_min) / (t_max - t_min)

print("Tableau normalisé :", temp_jour_normalisees.round(2))
print("Min après normalisation :", temp_jour_normalisees.min())  # 0
print("Max après normalisation :", temp_jour_normalisees.max())  # 1
```

## 4) Convertir le tableau initial en Fahrenheit 

### Formule de conversion Celsius → Fahrenheit

$$
F = \frac{9}{5}C + 32
$$

où $C$ représente la température en degrés Celsius et $F$ la température en degrés Fahrenheit.

```python
# 4) Conversion en Fahrenheit : F = C * 9/5 + 32
temp_jour_fahrenheit = temp_jour * (9 / 5) + 32
print("Températures en Fahrenheit :", temp_jour_fahrenheit.round(1))
```
